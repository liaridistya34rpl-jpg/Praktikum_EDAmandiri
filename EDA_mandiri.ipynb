```python
import pandas as pd

data_kantin = {
    'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
    'harga': [12000, 4000, 10000, 4000, 8000],
    'terjual': [23, 40, None, 35, 18]
}

df = pd.DataFrame(data_kantin)
print(df)
```

```python
import pandas as pd

df = pd.read_csv('data_kantin.csv')

print(df.head())
print(df.info())
print(df.describe())
print(df.shape)
```

```python
print(df.isnull().sum())

df['terjual'] = df['terjual'].fillna(0)

df = df.dropna(subset=['menu'])

print(df.isnull().sum())
```

```python
print(df.duplicated().sum())

df = df.drop_duplicates()

df['harga'] = df['harga'].astype(int)

df['terjual'] = df['terjual'].astype(int)

df['tanggal'] = pd.to_datetime(df['tanggal'])

print(df.dtypes)
print('Jumlah baris setelah dibersihkan:', len(df))
```

```python
# Filtering
laris = df[df['terjual'] > 20]
print("Menu dengan penjualan lebih dari 20:")
print(laris)

# Sorting
urut = df.sort_values(
    by='terjual',
    ascending=False
)
print("Data berdasarkan penjualan tertinggi:")
print(urut)

# Membuat kolom turunan
df['total_pendapatan'] = df['harga'] * df['terjual']

# Groupby
ringkasan = df.groupby('menu')['total_pendapatan'].sum()

print("Total pendapatan setiap menu:")
print(ringkasan.sort_values(ascending=False))
```
